```python
import tensorflow as tf
from tensorflow import keras
#from keras.models import load_model
#import argparse
import pickle
import cv2
import matplotlib.pyplot as plt
import os

model_path='C:/Users/betin/Desktop/Lab4_2-4_12/4_6/animals/output/model.keras'
print("[INFO] Загрузка НС...")
model = keras.models.load_model(model_path)
model.summary()

label_bin='C:/Users/betin/Desktop/Lab4_2-4_12/4_6/animals/output/net_lb.pickle'
print("[INFO] Загрузка меток...")
lb = pickle.loads(open(label_bin, "rb").read())

image_file='C:/Users/betin/Desktop/Lab4_2-4_12/4_6/images/cat3.jpg'

width=64
height=64
image = cv2.imread(image_file)
output = image.copy()
image = cv2.resize(image, (width, height))
# масштабируем значения пикселей к диапазону [0, 1]
image = image.astype("float") / 255.0
# и просто добавляем размер пакета
image = image.reshape((1, image.shape[0], image.shape[1],image.shape[2]))

print("[INFO] Предсказание...")
preds = model.predict(image)
print(preds)

print("[INFO] Индекс метки класса с наибольшей вероятностью...")
i = preds.argmax(axis=1)[0]
label = lb.classes_[i]
print(label)

output = cv2.resize(output, (356, 500))
print("[INFO] А это была..")
text = "{}: {:.2f}%".format(label, preds[0][i] * 100)
cv2.putText(output, text, (10, 30),
cv2.FONT_HERSHEY_SIMPLEX, 0.7,
(0, 0, 255), 2)
# показываем выходное изображение
# Конвертируем BGR (OpenCV) в RGB (matplotlib)
output_rgb = cv2.cvtColor(output, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(8, 8))
plt.imshow(output_rgb)
plt.axis('off')  # убираем оси
plt.title(f'Предсказание: {label} ({preds[0][i]*100:.2f}%)')
plt.show()

cv2.imwrite('C:/Users/betin/Desktop/Lab4_2-4_12/4_6/animals/images_test/result.jpg', output)
```
